# LIVER PATIENT DATA ANALYSIS PROJECT

```python
# Import Libraries

import pandas as pd
import numpy as np
```

```python
# Load Dataset

df = pd.read_csv("Liver Patient Dataset (LPD)_train.csv", encoding="latin1")
df.head()
```

```python
# Check Column Names

df.columns
```

# DATA CLEANING

```python
# Dataset Information

df.info()
```

```python
# Statistical Summary

df.describe()
```

```python
# Check Missing Values

df.isnull().sum()
```

```python
# Missing percentage

df.isnull().mean() * 100
```

```python
# Check Duplicate Records

df.duplicated().sum()
```

```python
# Remove

df = df.drop_duplicates()
```

```python
# Check

df.duplicated().sum()
```

```python
# Clean Column Names

df.columns = df.columns.str.strip()
```

```python
# check

df.columns
```

```python
# Clean Gender Column

df["Gender of the patient"] = df["Gender of the patient"].str.strip().str.title()
```

```python
# check

df["Gender of the patient"].value_counts()
```

```python
# Convert Numeric Columns

num_cols = df.columns[2:10]

df[num_cols] = df[num_cols].apply(pd.to_numeric, errors="coerce")
```

```python
# Handle Missing Numerical Values

df[num_cols] = df[num_cols].fillna(df[num_cols].median())
```

```python
# Handle Missing Gender

df["Gender of the patient"] = df["Gender of the patient"].fillna("Unknown")
```

```python
# Check Missing Values Again

df.isnull().sum()
```

```python
# Check Result Column

df["Result"].value_counts()
```

```python
# Final Dataset Check

df.shape
```

```python
df['Age of the patient'] = df['Age of the patient'].fillna(df['Age of the patient'].median())
```

 # Exploratory Data Analysis

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

Patient Result Distribution — Bar Chart

```python
result_count = df["Result"].value_counts()

plt.figure(figsize=(5,2))

plt.bar(result_count.index, result_count.values)

plt.title("Patient Result Distribution")
plt.xlabel("Result")
plt.ylabel("Count")

plt.show()
```

Age Distribution — Histogram

```python
plt.figure(figsize=(4,3))

plt.hist(df["Age of the patient"], bins=20)

plt.title("Age Distribution of Patients")
plt.xlabel("Age")
plt.ylabel("Count")

plt.show()
```

Gender Distribution — Bar Chart

```python
gender_count = df["Gender of the patient"].value_counts()

plt.figure(figsize=(5,2))

plt.bar(gender_count.index, gender_count.values)

plt.title("Gender Distribution of Patients")
plt.xlabel("Gender")
plt.ylabel("Count")

plt.show()
```

Total Bilirubin Distribution — Histogram

```python
plt.figure(figsize=(4,3))

plt.hist(df["Total Bilirubin"], bins=20)

plt.title("Total Bilirubin Distribution")
plt.xlabel("Total Bilirubin")
plt.ylabel("Count")

plt.show()
```

Direct Bilirubin Distribution — Histogram

```python
plt.figure(figsize=(4,3))

plt.hist(df["Direct Bilirubin"], bins=20)

plt.title("Direct Bilirubin Distribution")
plt.xlabel("Direct Bilirubin")
plt.ylabel("Count")

plt.show()
```

Average Liver Enzyme Levels — Bar Chart

```python
enzyme_names = [
    "Alkaline Phosphotase",
    "SGPT",
    "SGOT"
]

enzyme_values = [
    df["Alkphos Alkaline Phosphotase"].mean(),
    df["Sgpt Alamine Aminotransferase"].mean(),
    df["Sgot Aspartate Aminotransferase"].mean()
]

plt.figure(figsize=(4,3))

plt.bar(enzyme_names, enzyme_values)

plt.title("Average Liver Enzyme Levels")
plt.xlabel("Liver Enzyme")
plt.ylabel("Average Value")

plt.xticks(rotation=20)

plt.show()
```

Total Bilirubin vs Result — Box Plot

```python
plt.figure(figsize=(4,3))

df.boxplot(column="Total Bilirubin", by="Result")

plt.title("Total Bilirubin vs Patient Result")
plt.xlabel("Result")
plt.ylabel("Total Bilirubin")

plt.suptitle("")
plt.show()
```

Age vs Result — Box Plot

```python
result_categories = df["Result"].unique()

bilirubin_data = []

for result in result_categories:
    bilirubin_data.append(
        df[df["Result"] == result]["Total Bilirubin"]
    )

plt.figure(figsize=(4,3))

plt.boxplot(
    bilirubin_data,
    tick_labels=result_categories
)

plt.title("Total Bilirubin vs Patient Result")
plt.xlabel("Result")
plt.ylabel("Total Bilirubin")

plt.show()
```

Correlation Heatmap

```python
correlation = df.select_dtypes(include="number").corr()

plt.figure(figsize=(5,3))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

Albumin Distribution — Histogram

```python
plt.figure(figsize=(4,3))

plt.hist(df["ALB Albumin"], bins=20)

plt.title("Albumin Distribution")
plt.xlabel("Albumin")
plt.ylabel("Count")

plt.show()
```

Patient Result by Gender — Bar Chart

```python
gender_result = pd.crosstab(
    df["Gender of the patient"],
    df["Result"]
)
x = np.arange(len(gender_result.index))
width = 0.35
plt.figure(figsize=(4,3))
plt.bar(
    x - width/2,
    gender_result.iloc[:, 0],
    width,
    label=str(gender_result.columns[0])
)
plt.bar(
    x + width/2,
    gender_result.iloc[:, 1],
    width,
    label=str(gender_result.columns[1])
)
plt.title("Patient Result by Gender")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.xticks(x, gender_result.index)
plt.legend()
plt.show()
```

Average Bilirubin by Result

```python
bilirubin_average = df.groupby("Result")["Total Bilirubin"].mean()

plt.figure(figsize=(5,3))

plt.bar(
    bilirubin_average.index,
    bilirubin_average.values
)

plt.title("Average Total Bilirubin by Result")
plt.xlabel("Result")
plt.ylabel("Average Total Bilirubin")

plt.show()
```

Gender Distribution — Pie Chart

```python
gender_count = df["Gender of the patient"].value_counts()

plt.figure(figsize=(5,5))

plt.pie(
    gender_count.values,
    labels=gender_count.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Gender Distribution of Patients")

plt.show()
```
